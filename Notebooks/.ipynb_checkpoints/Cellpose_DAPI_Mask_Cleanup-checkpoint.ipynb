```python
# Data Loading --> Loading folder content

DAPI_mask_data = glob(os.path.join(folder_path, '*DAPI SEM_seg.npy'))
DAPI_mask_data.sort()


for mask_path in DAPI_mask_data:
    print(os.path.basename(mask_path))
```

```python
# Cellpose DAPI Mask Cleanup High Throughput
# This script cleans up masks that were generated by cellpose by removing all the 
# nuclei that touch the border of the picture and all fragments below a size threshold

for i in range (len(DAPI_mask_data)):
    # Try to load as dictionary (original Cellpose format) or plain array (already cleaned)
    loaded_data = np.load(DAPI_mask_data[i], allow_pickle=True)
    
    # Try to load it as an output from Cellpose
    try:
        mask_dict = loaded_data.item()
        mask = mask_dict['masks']
    # If it isn't an output from Cellpose, but a plain array
    except (ValueError, AttributeError):
        mask = loaded_data

    # Calculate the properties (area etc.) for the nuclei
    props = measure.regionprops(mask)
    
    
## Removal of the Nuclei that touch the border of the picture
    # Array to later save the nuclei that don't touch the border
    labels_to_keep = []
    for region in props:
        # Getting the coordinates for the borders of the picture
        min_row, min_col, max_row, max_col = region.bbox
        # Checking if the nuclei touch any of the borders
        touches_border = (min_row == 0 or min_col == 0 or 
                          max_row == 2048 or max_col == 2048)
    
        if not touches_border:
            # Save the nuclei that don't touch the border in the above created array
            labels_to_keep.append(region.label)

    # Create a new mask without the nuclei that touch the border as a starting point for the next step
    mask_clean_border = np.zeros_like(mask)

    for label in labels_to_keep:
        # Copy the labels for the nuclei from the old mask to the new one
        mask_clean_border[mask == label] = label
    # Get the properties from the nuclei on the new mask
    props_no_border = measure.regionprops(mask_clean_border)

## Small fragment cleanup
    # Minimum size for a Nucleus in pixels
    threshold = 1000  

    # Array for saving cleared nuclei later
    labels_final = []
    for region in props_no_border:
        # Keeping only the nuclei that clear the earlier thershold
        if region.area >= threshold:
            labels_final.append(region.label)
    
    # Create final clean mask
    clean_mask = np.zeros_like(mask)
    for label in labels_final:
        # Copy the Nuclei that passed both requirements
        clean_mask[mask_clean_border == label] = label


    # Create an empty array with the same dimensions like the clean_mask
    clean_mask_renumbered = np.zeros_like(clean_mask)

    # Get the unique labels (e.g., [0, 3, 7, 12, 15]) from the clena_mask
    unique_labels = np.unique(clean_mask)
    # Removing the Background from the labels (Background = 0)
    unique_labels = unique_labels[unique_labels != 0]
    
    # Renumber: 3→1, 7→2, 12→3, 15→4
    for new_label, old_label in enumerate(unique_labels, start=1):
        clean_mask_renumbered[clean_mask == old_label] = new_label



    # Extract the name of the picture for saving
    base_name = os.path.basename(DAPI_mask_data[i])

    # Saving the cleaned up mask in the same folder where it originated from with the same name (overwriting the old one)
    np.save(folder_path+'/'+base_name, clean_mask_renumbered, allow_pickle=True)
```

```python
# Small fragment cleanup

# Remove nuclei smaller than threshold
threshold = 1000  # Set your threshold based on the histogram

labels_final = []
for region in props_no_border:
    if region.area >= threshold:
        labels_final.append(region.label)

# Create final clean mask
clean_mask = np.zeros_like(mask)
for label in labels_final:
    clean_mask[mask_clean_border == label] = label



#Get properties of remaining nuclei and plot sizes
props_no_border = measure.regionprops(clean_mask)
areas = [region.area for region in props_no_border]

plt.figure(figsize=(10, 6))
plt.hist(areas, bins=50, edgecolor='black')
plt.xlabel('Nucleus size (pixels)')
plt.ylabel('Count')
plt.title('Size distribution of nuclei (small fragments removed)')
plt.show()
```
