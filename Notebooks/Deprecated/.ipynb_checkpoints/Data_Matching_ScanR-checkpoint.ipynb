```python
import pandas as pd
import numpy as np
from scipy.spatial.distance import cdist
from tqdm import tqdm
```

```python
# Define function that does everything
def match_nuclei_datasets(dataset_Cellpose_path, dataset_ScanR_path, tolerance=10, 
                          x_col_cellpose='nucleus_centroid_x', y_col_cellpose='nucleus_centroid_y',
                          x_col_ScanR='nucleus_centroid_x', y_col_ScanR='nucleus_centroid_y', 
                          well_col_scanr='Well ', position_col_scanr='Position ', well_col_cellpose='Well', 
                          position_col_cellpose='Position', verbose=False, show_warnings=True, save_details_to=None):
    """
    Match nuclei between two datasets based on centroid proximity within matching Well/Position groups.
    
    Parameters:
    -----------
    dataset_Cellpose_path : str
        Path to your program's CSV file
    dataset_ScanR_path : str
        Path to commercial software's CSV file
    tolerance : float
        Maximum distance (in pixels) for nuclei to be considered a match
    x_col_cellpose, y_col_cellpose : str
        Column names for x,y coordinates in Cellpose dataset
    x_col_ScanR, y_col_ScanR : str
        Column names for x,y coordinates in ScanR dataset
    well_col_scanr, position_col_scanr, well_col_cellpose, position_col_cellpose : str
        Column names for Well and Position identifiers
    verbose : bool (default=False)
        If True, print details for each image. If False, only show progress bar and summary.
    show_warnings : bool (default=True)
        If True, print warnings for images with issues (no cells, low match rate, etc.)
    save_details_to : str or None (default=None)
        If provided, save per-image statistics to this CSV file path
    
    Returns:
    --------
    pd.DataFrame
        Combined DataFrame with matched and unmatched nuclei, grouped by Well/Position
    """
    
    # Load both datasets
    # Trying to load the data from cellpose with skipping the first line at first to skip the sep=, line and if that was removed/doesn't exist 
    # after saving it once in excel try to load the .csv file with no line skipping and sep=;
    try:
        df_cellpose  = pd.read_csv(dataset_Cellpose_path, skiprows=1, header = 0)
        # Add source labels
        df_cellpose ['source'] = 'Cellpose'
        # Get unique Well/Position combinations from YOUR dataset (smaller one)
        cellpose_groups = df_cellpose[[well_col_cellpose, position_col_cellpose]].drop_duplicates()
    except:
        df_cellpose  = pd.read_csv(dataset_Cellpose_path, skiprows=0, sep=';', header = 0)
        # Add source labels
        df_cellpose ['source'] = 'Cellpose'
        # Get unique Well/Position combinations from YOUR dataset (smaller one)
        cellpose_groups = df_cellpose[[well_col_cellpose, position_col_cellpose]].drop_duplicates()
        
    df_scanr  = pd.read_csv(dataset_ScanR_path, sep=';')
    # Add source labels
    df_scanr ['source'] = 'ScanR'
    
    # Filter ScanR dataset to only include matching Well/Position combinations
    # Rename cellpose columns to match ScanR columns
    cellpose_groups_for_merge = cellpose_groups.rename(columns={
        well_col_cellpose: well_col_scanr,
        position_col_cellpose: position_col_scanr
    })
    
    # Merge using the ScanR column names
    df_scanr_filtered = df_scanr.merge(
        cellpose_groups_for_merge,
        on=[well_col_scanr, position_col_scanr],  
        how='inner'
    )


    print(f"\nOriginal ScanR dataset: {len(df_scanr)} cells")
    print(f"Filtered ScanR dataset: {len(df_scanr_filtered)} cells (matching Well/Position)")
    print(f"Cellpose dataset: {len(df_cellpose)} cells")
    print(f"\nProcessing {len(cellpose_groups)} unique Well/Position combinations...")
    
    # Store results for each group
    all_result_rows = []
    per_image_stats = []  # For optional CSV export

    # Use tqdm for progress bar (disable if verbose mode is on)
    iterator = tqdm(cellpose_groups.iterrows(), total=len(cellpose_groups), 
                   desc="Matching nuclei", disable=verbose)
    
    # saves row = [well, Position] for each unique dataset
    for idx, row in iterator:
        well = row[well_col_cellpose]
        position = row[position_col_cellpose]
        # Filter both datasets for this specific Well/Position, by creating a copy that only contains these
        df1_group = df_cellpose[
            (df_cellpose[well_col_cellpose] == well) & 
            (df_cellpose[position_col_cellpose] == position)
        ].copy()
        
        df2_group = df_scanr_filtered[
            (df_scanr_filtered[well_col_scanr] == well) & 
            (df_scanr_filtered[position_col_scanr] == position) &
            (df_scanr_filtered['Area '] > 1000)  # Remove all the datapoints from the ScanR datasheet that have an area of under 1000 to match the cellpose dataset
        ].copy()


        n_cellpose = len(df1_group)
        n_scanr = len(df2_group)

        if verbose:
            print(f"\nWell {well}, Position {position}:")
            print(f"  Cellpose: {len(df1_group)} cells")
            print(f"  ScanR: {len(df2_group)} cells")
            
        if n_cellpose == 0:
            if show_warnings:
                print(f"⚠️  Well {well}, Position {position}: No Cellpose cells found")
            continue
            
        if n_scanr == 0:
            if show_warnings:
                print(f"⚠️  Well {well}, Position {position}: No ScanR cells found ({n_cellpose} Cellpose cells unmatched)")
            
            # All Cellpose cells are unmatched for this group
            for i in range(n_cellpose):
                row_data = df1_group.iloc[i].to_dict()
                combined_row = {
                    'status': 'UNCONFIRMED',
                    'match_distance': np.nan,
                    'source': 'Cellpose_only',
                    'Well': well,
                    'Position': position,
                    **{f'd_Cellpose_{k}': v for k, v in row_data.items() if k not in ['source', well_col_cellpose, position_col_cellpose]},
                }
                all_result_rows.append(combined_row)
                
            # Save stats
            per_image_stats.append({
                'Well': well, 'Position': position,
                'Cellpose_cells': n_cellpose, 'ScanR_cells': n_scanr,
                'Matches': 0, 'Match_rate_%': 0,
                'Unmatched_Cellpose': n_cellpose, 'Unmatched_ScanR': 0
            })
            continue

        
        # Extract coordinates for this group
        coords_cellpose = df1_group[[y_col_cellpose, x_col_cellpose]].values
        coords_scanr = df2_group[[y_col_ScanR, x_col_ScanR]].values
        
        # Calculate pairwise distances between all nuclei
        distances = cdist(coords_cellpose, coords_scanr)
        
        # Find matches within tolerance
        matched_pairs = []
        matched_idx1 = set()
        matched_idx2 = set()
        
        # For each nucleus in Cellpose, find closest match in ScanR
        for i in range(len(coords_cellpose)):
            # Finds the closest nucleus in ScanR (: means all from that row) to the one selected with i in the cellpose dataset
            min_dist_idx = np.argmin(distances[i, :])
            # min_dist is that smallest value between the two datasets for i in cellpose
            min_dist = distances[i, min_dist_idx]
            
            if min_dist <= tolerance:
                # Check if this is also the closest match from ScanR's perspective
                if np.argmin(distances[:, min_dist_idx]) == i:
                    matched_pairs.append({
                        'idx1': i,
                        'idx2': min_dist_idx,
                        'distance': min_dist
                    })
                    matched_idx1.add(i)
                    matched_idx2.add(min_dist_idx)
                    
        n_matches = len(matched_pairs)
        match_rate = (n_matches / n_cellpose * 100) if n_cellpose > 0 else 0
        
        # Warning for low match rates
        if show_warnings and match_rate < 50 and n_cellpose >= 10:
            print(f"⚠️  Well {well}, Position {position}: Low match rate ({match_rate:.1f}% - {n_matches}/{n_cellpose})")

        # Save per-image statistics
        per_image_stats.append({
            'Well': well,
            'Position': position,
            'Cellpose_cells': n_cellpose,
            'ScanR_cells': n_scanr,
            'Matches': n_matches,
            'Match_rate_%': match_rate,
            'Avg_match_distance': np.mean([m['distance'] for m in matched_pairs]) if matched_pairs else np.nan,
            'Unmatched_Cellpose': n_cellpose - n_matches,
            'Unmatched_ScanR': n_scanr - n_matches
        })
        
        # Add CONFIRMED nuclei (matched)
        for match in matched_pairs:
            row1 = df1_group.iloc[match['idx1']].to_dict()
            row2 = df2_group.iloc[match['idx2']].to_dict()
            
            combined_row = {
                'status': 'CONFIRMED',
                'match_distance': match['distance'],
                'source': 'Both',
                'Well': well,
                'Position': position,
                **{f'd_Cellpose_{k}': v for k, v in row1.items() if k not in ['source', well_col_cellpose, position_col_cellpose]},
                **{f'd_ScanR_{k}': v for k, v in row2.items() if k not in ['source', well_col_scanr, position_col_scanr]}
            }
            all_result_rows.append(combined_row)
        
        # Add UNCONFIRMED nuclei from Cellpose
        unmatched1 = [i for i in range(len(df1_group)) if i not in matched_idx1]
        for i in unmatched1:
            row_data = df1_group.iloc[i].to_dict()
            combined_row = {
                'status': 'UNCONFIRMED',
                'match_distance': np.nan,
                'source': 'Cellpose_only',
                'Well': well,
                'Position': position,
                **{f'd_Cellpose_{k}': v for k, v in row_data.items() if k not in ['source', well_col_cellpose, position_col_cellpose]},
            }
            all_result_rows.append(combined_row)
        
        # Add UNCONFIRMED nuclei from ScanR
        unmatched2 = [i for i in range(len(df2_group)) if i not in matched_idx2]
        for i in unmatched2:
            row_data = df2_group.iloc[i].to_dict()
            combined_row = {
                'status': 'UNCONFIRMED',
                'match_distance': np.nan,
                'source': 'ScanR_only',
                'Well': well,
                'Position': position,
                **{f'd_ScanR_{k}': v for k, v in row_data.items() if k not in ['source', well_col_scanr, position_col_scanr]},
            }
            all_result_rows.append(combined_row)
    
    # Create final DataFrame
    result_df = pd.DataFrame(all_result_rows)
    
    # Sort by Well, Position, and status (CONFIRMED first, then UNCONFIRMED)
    #Adds a column with 0 for confirmed and 1 for unconfirmed to sort easier
    status_order = {'CONFIRMED': 0, 'UNCONFIRMED': 1} 
    result_df['status_order'] = result_df['status'].map(status_order)
    result_df = result_df.sort_values(['Well', 'Position', 'status_order', 'source'])
    result_df = result_df.drop('status_order', axis=1)

    # Save per-image details if requested
    if save_details_to:
        # Write sep=, line for Excel compatibility
        with open(save_details_to, 'w') as f:
            f.write('sep=,\n')
        stats_df = pd.DataFrame(per_image_stats)
        stats_df.to_csv(save_details_to, mode='a', header=True, index=False)
        print(f"\n📊 Per-image statistics saved to: {save_details_to}")
    
    # Print overall summary
    print("\n" + "="*60)
    print("OVERALL SUMMARY")
    print("="*60)
    print(f"Total Cellpose cells: {len(df_cellpose)}")
    print(f"Total ScanR cells (filtered): {len(df_scanr_filtered)}")
    print(f"Total confirmed matches: {len(result_df[result_df['status'] == 'CONFIRMED'])}")
    print(f"Total unmatched Cellpose: {len(result_df[result_df['source'] == 'Cellpose_only'])}")
    print(f"Total unmatched ScanR: {len(result_df[result_df['source'] == 'ScanR_only'])}")
    print(f"Total rows in result: {len(result_df)}")

    # Calculate and display overall statistics
    if per_image_stats:
        stats_df = pd.DataFrame(per_image_stats)
        avg_match_rate = stats_df['Match_rate_%'].mean()
        print(f"\nAverage match rate: {avg_match_rate:.1f}%")
        print(f"Match rate range: {stats_df['Match_rate_%'].min():.1f}% - {stats_df['Match_rate_%'].max():.1f}%")
        
        avg_dist = stats_df['Avg_match_distance'].mean()
        if not np.isnan(avg_dist):
            print(f"Average match distance: {avg_dist:.2f} pixels")
            
    print("="*60)
    
    return result_df

```

```python


# Call on function with the specific parameters
matched_df = match_nuclei_datasets(
    dataset_Cellpose_path=
    r'Y:\Group Members\Valentin Aubry\01_Data\AP470_U2OS_gH2AX(488)_53BP1(568)_EdU(647)_20x_002\data\3_param_iteration--nuclei_DF_SEM.csv',
    dataset_ScanR_path=
    r'Y:\Group Members\Valentin Aubry\01_Data\AP470_U2OS_gH2AX(488)_53BP1(568)_EdU(647)_20x_002\AP470_U2OS_gH2AX(488)_53BP1(568)_EdU(647)_20x_002_xy_positions_Main.csv',
    tolerance=30,
    x_col_cellpose='centr_x',  
    y_col_cellpose='centr_y',
    x_col_ScanR='Center X ',  
    y_col_ScanR='Center Y ',
    save_details_to = r'Y:\Group Members\Valentin Aubry\01_Data\AP470_U2OS_gH2AX(488)_53BP1(568)_EdU(647)_20x_002\data\1_match_stats.csv'
)

filename = r'Y:\Group Members\Valentin Aubry\01_Data\AP470_U2OS_gH2AX(488)_53BP1(568)_EdU(647)_20x_002\data\1_cellpose_vs_ScanR.csv'

# Write sep=, line for Excel compatibility
with open(filename, 'w') as f:
    f.write('sep=,\n')

# Write data with header
matched_df.to_csv(filename, mode='a', header=True, index=False)
```
